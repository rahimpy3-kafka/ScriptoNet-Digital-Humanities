## Candidate Selection Strategy

Remaining passages are ranked using emotional and reflective linguistic indicators.

The scoring system identifies passages containing potential therapeutic elements such as:

- emotional expression
- memory and reflection
- interpersonal support
- acceptance
- transformation

```python
therapeutic_keywords = [
    "memory",
    "remember",
    "reflection",
    "reflect",
    "feel",
    "feeling",
    "heart",
    "mind",
    "sorrow",
    "grief",
    "hope",
    "comfort",
    "friend",
    "sympathy",
    "love",
    "affection",
    "forgive",
    "forgiveness",
    "peace",
    "happiness",
    "joy",
    "understand",
    "learn",
    "accept",
    "calm",
    "consolation",
    "support"
]

print("Keyword list created")
```

```python
def keyword_score(text):
    text = text.lower()

    score = 0

    for word in therapeutic_keywords:
        if word in text:
            score += 1

    return score


remaining_df["keyword_score"] = remaining_df["text"].apply(
    keyword_score
)

remaining_df.head()
```

```python
candidate_df = remaining_df.sort_values(
    by="keyword_score",
    ascending=False
)

candidate_df[
    ["passage_id", "keyword_score", "text"]
].head(20)
```
