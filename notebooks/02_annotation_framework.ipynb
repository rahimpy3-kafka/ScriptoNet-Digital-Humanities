# Annotation Framework

## ScriptoNet Digital Humanities Project

This notebook defines the annotation framework used to transform literary interpretation into structured data for Artificial Intelligence models.

The purpose of annotation is to create examples where literary passages are connected with meaningful categories that a machine learning model can learn from.

## Why Annotation Matters

A computer model cannot directly understand literary meaning.

For example:

A passage may contain:
- letters
- memories
- emotions
- suffering

However, these elements do not automatically indicate a therapeutic process.

Human interpretation is required to identify whether a passage represents:
- emotional processing
- healing
- reflection
- transformation

Annotation creates the bridge between literary analysis and machine learning.

## Annotation Task 1: Therapeutic Representation Detection

Each passage receives a binary label.

Label 1:
The passage contains a representation of emotional processing, reflection, healing, or transformation.

Label 0:
The passage does not contain a therapeutic representation.

This creates the first supervised learning task.

# Annotation Guidelines

Annotation guidelines define how literary passages will be labelled for machine learning.

The purpose is to ensure that human interpretation follows consistent rules before creating training data.

## Annotation Level 1: Therapeutic Representation

Question:

Does this passage represent a process of emotional processing, healing, reflection, or transformation?

Labels:

1 = Present

0 = Not Present


Label 1 examples:

- A character reflects on suffering and develops new understanding.
- Writing is used to process experiences.
- Emotional expression leads to transformation.


Label 0 examples:

- Physical descriptions
- Travel information
- Simple events without emotional processing
- Narrative background information

## Annotation Level 2: Therapeutic Mechanism

If the therapeutic representation is present, assign one category:

### 1. Scriptotherapy

Writing, letters, journals, confession, or storytelling used for emotional processing.


### 2. Poetry-mediated Processing

Poetry used for emotional expression or reflection.


### 3. Music-mediated Processing

Music used for emotional access, memory, or regulation.


### 4. Narrative Reconstruction

A character changes their understanding of themselves, their past, or their experiences.


### 5. Emotional Catharsis

Release or expression of previously contained emotions.

## Annotation Level 3: Intensity

Rate the strength of therapeutic representation.

1 = Very weak

2 = Weak

3 = Moderate

4 = Strong

5 = Very strong
