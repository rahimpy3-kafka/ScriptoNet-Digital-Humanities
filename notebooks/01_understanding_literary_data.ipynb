## Loading Literary Text Data

In this section, we load a public domain literary text and prepare it for computational analysis.

The purpose is to understand how a literary work can be transformed into a structured dataset before applying Natural Language Processing methods.

```python
import requests

url = "https://www.gutenberg.org/files/84/84-0.txt"

response = requests.get(url)

text = response.text

print(text[:1000])
```

## Basic Text Exploration

Before applying Natural Language Processing methods, we first examine the basic properties of the literary text.

This helps us understand the size, structure, and characteristics of the dataset.

```python
# Number of characters
characters = len(text)

# Split text into words
words = text.split()

# Number of words
word_count = len(words)

print("Total characters:", characters)
print("Total words:", word_count)
```

## Observation

The literary text has been successfully converted into digital form.

The first exploration shows:

- Total characters: 419,434
- Total words: 75,060

At this stage, the computer only understands the text as a sequence of characters and words.

However, literary meaning such as trauma, healing, memory, and emotional transformation is not directly available from raw text.

The next challenge is to transform this continuous text into meaningful computational units while preserving enough context for literary analysis.

## Passage Segmentation

Machine learning models cannot directly analyse an entire novel as a single unit.

Therefore, the literary text is divided into smaller passages.

Each passage becomes a potential research sample that can later be annotated and used for NLP experiments.

The goal is to preserve enough context for literary interpretation while creating manageable computational units.

```python
# Split text into paragraphs

paragraphs = text.split("\n\n")

# Remove empty paragraphs
paragraphs = [
    p.strip()
    for p in paragraphs
    if p.strip()
]

print("Total paragraphs:", len(paragraphs))

# Show first paragraph
print(paragraphs[0])
```


## Observation: Text Cleaning Requirement

The raw text contains non-literary information such as Project Gutenberg metadata.

Example:

"*** START OF THE PROJECT GUTENBERG EBOOK 84 ***"

This information is not part of the literary narrative and should be removed before computational analysis.

This demonstrates an important challenge in Digital Humanities:

Computers process all available text, but researchers must identify which parts contain meaningful literary information.

## Removing Non-Literary Metadata

Before analysing literary patterns, we remove unnecessary text markers from the digital source.

The goal is to keep only the actual narrative content.

```python
# Remove Gutenberg header and footer information

start_marker = "*** START OF THE PROJECT GUTENBERG EBOOK 84 ***"
end_marker = "*** END OF THE PROJECT GUTENBERG EBOOK 84 ***"

clean_text = text.split(start_marker)[1]
clean_text = clean_text.split(end_marker)[0]

clean_text = clean_text.strip()

print(clean_text[:1000])
```

## Observation: Literary Structure Matters

After removing metadata, the text reveals its original narrative structure.

Frankenstein begins with a series of letters before moving into the main narrative.

This demonstrates an important challenge for computational literary analysis:

The same textual form can have different meanings depending on context.

For example:

A letter may represent:
- simple communication
- narrative framing
- emotional reflection
- confession or therapeutic processing

Therefore, AI models must learn contextual meaning rather than only detecting keywords.

## Creating Computational Passages

Machine learning models require structured examples.

A complete novel is too large and complex to use directly.

Therefore, the text is divided into smaller passages.

Each passage becomes an individual unit for future annotation and AI analysis.

The goal is to maintain enough literary context while creating manageable data samples.

```python
# Create passages from cleaned paragraphs

paragraphs = [
    p.strip()
    for p in clean_text.split("\n\n")
    if p.strip()
]

print("Number of cleaned paragraphs:", len(paragraphs))


# Combine paragraphs into larger passages

passages = []

current_passage = ""

for paragraph in paragraphs:
    current_passage += " " + paragraph

    # approximately create passages of 200-300 words
    if len(current_passage.split()) >= 250:
        passages.append(current_passage.strip())
        current_passage = ""

# Add remaining text
if current_passage:
    passages.append(current_passage.strip())


print("Total passages created:", len(passages))
```

## Observation: From Literature to Dataset

The literary text has been transformed into structured computational units.

The original novel contained continuous narrative flow. For computational analysis, it was divided into 229 passages.

Each passage can now become an independent research sample.

However, segmentation alone does not create understanding.

A computer can identify passages, but it does not know whether a passage represents:

- trauma
- healing
- reflection
- confession
- emotional transformation

Human annotation is required to provide meaningful interpretation.

```python
import pandas as pd

# Create dataframe

df = pd.DataFrame({
    "passage_id": range(1, len(passages)+1),
    "text": passages
})

df.head()
```

```python
df.to_csv(
    "frankenstein_passages.csv",
    index=False
)

print("Dataset saved successfully")
```
